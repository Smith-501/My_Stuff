A **fractal** is an image repeated on an ever-reduced scale

###The Cantor Set Fractal

```python
initializeTurtle()
def cantor(x, y, length):
  speed = 13 #fastest speed of turtle
  if length >= 5:
    penup()
    pensize(2)
    pencolor('blue')
    setpos(x, y)
    pendown()
    fd(length) #fd = forward
    y -= 80
    cantor(x, y, length/3)
    cantor(x + 2 * length/3, y, length/3)
    penup()
    setpos(x, y + 80)
cantor(-400, 200, 500)
```

###The Koch Snowflake

```python
initializeTurtle()
pensize(1)
penup()
setpos(-250, 200)
pendown()
def KochSnowflake(length,level):
  speed(13)
  for i in range(3):
    plot_side(length, level)
    rt(120) #rt = rigth turn
def plot_side(length, level):
  if level == 0:
    fd(length)
    return
  plot_side(length/3, level-1)
  lt(60)
  plot_side(length/3, level-1)
  rt(120)
  plot_side(length/3, level-1)
  lt(60)
  plot_side(length/3, level-1)
KochSnowflake(300, 3)
```

###Bifurcating Tree

```python
initializeTurtle()
speed(13)
setheading(90)
penup()
setpos(-200, -120)
pendown()
def FractalTreeColor(length, level):
  pensize(length/10)
  if length < 20:
    pencolor('green')
  else:
    pencolor('brown')
  if level > 0:
    fd(length)
    rt(60)
    FractalTreeColor(length*0.7, level-1)
    lt(90)
    FractalTreeColor(length*0.5, level-1)
    rt(30)
    penup()
    bk(length)
    pendown()
FractalTreeColor(200, 10)
```

###The Sierpinski Triangle

```python
initializeTurtle()
speed(13)
penup()
setpos(-300, 0)
pendown()
def SierpinksiTriangle(length, level):
  if level==0:
    return
  begin_fill()
  color('red')
  for i in range(3):
    SierpinksiTriangle(length/2, level-1)
    fd(length)
    lt(120)
  end_fill()
SierpinksiTriangle(300, 5)
```

1. Plot a variation of the Cantor set, where the two middle segments
are removed at each stage. Thus, at each stage, one segment is replaced with three segments each on-fifth the length of the previous
segment.
####

```python
initializeTurtle()
def cantor3(x, y, length):
  speed = 13 #fastest speed of turtle
  if length >= 1:
    penup()
    pensize(2)
    pencolor('blue')
    setpos(x, y)
    pendown()
    fd(length) #fd = forward
    y -= 50
    cantor3(x, y, length/5)
    cantor3(x + 2 * length/5, y, length/5)
    cantor3(x + 4 * length/5, y, length/5)
    penup()
    setpos(x, y + 50)
cantor(-400, 200, 500)
```

### 4.4 Surface Plots

Pot the surface of $z = x^2 + y^2$

```python
!pip install plotly
```

```python
import numpy as np
import plotly.graph_objects as go

# Create mesh grid
x = np.linspace(-4, 4, 100)
y = np.linspace(-4, 4, 100)
X, Y = np.meshgrid(x, y)
# Calculate surface
Z = X**2 + Y**2
# Create figure
fig = go.Figure()
# Coloured surface
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z,
    colorscale='Jet',
    opacity=1,
    showscale=True,
    colorbar=dict(title='Height')
))
# Black wireframe
for i in range(0, len(x), 2):
    fig.add_trace(go.Scatter3d(
        x=X[i, :],
        y=Y[i, :],
        z=Z[i, :],
        mode='lines',
        line=dict(color='black', width=2),
        showlegend=False
    ))
for j in range(0, len(y), 2):
    fig.add_trace(go.Scatter3d(
        x=X[:, j],
        y=Y[:, j],
        z=Z[:, j],
        mode='lines',
        line=dict(color='black', width=2),
        showlegend=False
    ))
# Layout
fig.update_layout(
    title='3D Surface with Wireframe',
    width=850,
    height=700,
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='cube',
        bgcolor='white'
    )
)
fig.show()
```

###6.4 Differentiation and Integretion (Calculus)

```python
x, y, c = symbols('x y c')
xval = -1/2
y = 4*x**3 - 3*x**4
dydx = diff(y, x)
m = dydx.subs(x, xval)
print('m = ', m)
yval = y.subs(x, xval)
print('y-value = ', yval)
c = yval - m*xval
print('c = ', c)
x = np.linspace(-1, 1.5, 100)
y= 4*x**3 - 3*x**4
plt.rcParams['font.size'] = 14
plt.plot(x, y, label = r'$y=4x^3-3x^4$')
plt.plot(x, m*x+c, label = r'$y=mx+c$')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
